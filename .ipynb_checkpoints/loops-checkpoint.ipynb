```python
arr = [1, 2, 3, 4, 3, 5, 3, 2]
target = 3

i = 0
while i < len(arr):
    if arr[i] != target:
        print(arr[i], end=" ")
    i += 1
```

```python
## Remove Target Element from list of integers
arr=list(map(int,input().split()))
target=int(input())
res=[]
for num in arr:
    if num!=target:
        res.append(num)
print(res)
         
```

```python
arr=list(map(int,input().split()))
target=int(input())
res=[]
n=len(arr)
for i in range(n):
    if arr[i]!=target:
        res.append(arr[i])
print(res)
         
```

```python
arr=list(map(int,input().split()))
target=int(input())
res=[]
n=len(arr)
i=0 ##initialization
while i<n:
    if arr[i]!=target:
        res.append(arr[i])
    i+=1
print(res)
```

```python
arr = [1, 2, 3, 4, 3, 5, 3]
res = []
for num in arr:
    if num not in res:
        res.append(num)

print(*res, sep=",")
```

```python
### Control Statement

arr=[1,9,5,6,9]
target=5
for i in range(len(arr)):
    if arr[i]==target:
        print(i)
        break
else:
    print(-1)
```
