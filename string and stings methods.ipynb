### String Methods
- Case Coversion Methods
 - Upper():It coverts entire stirng into uppercase
 - Lower():It converts entire stirng into lowercase
 - title():It coverts every word first character into uppercase and remaining into lowercase
 - swapcase():It coverts lower into upper and to lowercase
 - Captalize():It converts first chracter of string into uppercase and remaining into lowercase

```python
string = "The ocean is blue and sky is BLUE"
lower = string.lower()
print(lower)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
string = "The ocean is blue and sky is BLUE72728@"
lower = string.lower()
print(lower)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
# white space remove methods
# strip():It removes the white space to both sides of string
# lstrip():It removes the leading(left) white space  of the string
# rstrip():It removes the lagging(right) white space  of the string
string="          codegnan           "
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
# white space remove methods
# strip():It removes the white space to both sides of string
# lstrip():It removes the leading white space  of the string
# rstrip():It removes the lagging white space  of the string
string="@@@@@@@@@code  @  gnan@@@@@@@@@"
print(string)
print(string.strip("@"))
print(string.lstrip("@"))
print(string.rstrip())
```

```python
# split(): It returns the list of substring and splits at white space
string = "The ocean is blue and sky is BLUE 72728@"
print(string.split())
print(string.split('is'))
print(string.split('z'))
print(string.split('blue'))
print(string.split('@'))
```

```python
#index (substring) methods
#Index(sub_string) parameters: It returns the first occurance of substring  if exists, otherwise "value error"
#find(sub_String):It returns the first occurance of substring  if exists, otherwise "-1"
string = "The ocean is blue and sky is blue"
print(string.index("o"))
print(string.index("is"))
print(string.index("is", 14))# t starts checking from 14 Index
print(string.index("z"))  ##ValueError: substring not found
```

```python
#index finding methods
#Index(sub_string) parameters: It returns the first occurance of substring  if exists, otherwise "value error"
#find(sub_String):It returns the first occurance of substring  if exists, otherwise "-1"
string = "The ocean is blue and sky is blue"
print(string.find("o"))
print(string.find("is"))
print(string.find("is", 14))# t starts checking from 14 Index
print(string.find("z"))
print(string.find(" "))
```

```python
# Replace(old with new ):It replace the old_string with new_string in string
string = "The ocean is blue and sky is blue"
print(string.replace('blue','red'))
print(string.replace('blue','red',1))#It replaces only 1 occurance
print(string.replace('blue','red',-1))
```

```python
# "sep".join(iterable): it joins the all iterable elements with seperators
lst=['The', 'ocean', 'is', 'blue', 'and', 'sky', 'is', 'BLUE']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

```python
####string checking methods 
-islower():It returns Ture when entire string is in lowercase
-isupper():It returns Ture when entire string is in uppercase
-istitle():It returns True when string is title format
-isalpha():It returns True when string contains Alphabets(i.e lower and uppercase)
-isalnum():It returns True when string either Alphabets or digits
-isdigit():It returns True when string contains only digits
-startswith(substring):It returns True when string startswith the sub string
-endwith(substring): It returns true when string endstwith the sub string
```

```python
print("hello".islower())
print("hELLO".isupper())
print("Hello".istitle())
print("223".isdigit())
print("abdABD".isalpha())
print("123abc".isalnum())
print("83873hdj".isalpha())
print("Hello World".startswith('H'))
print("HELLO world".endswith('d'))
```
