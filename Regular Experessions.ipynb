## REGULAR EXPRESSION


```python
import re
```

```python
## matc(pat, string)
string= "The sky is blue and Ocean is blue"
res=re.match("[A-Za-z]", string)
print(res)
print(res.span()) ##it returns the match positions
print(res.group()) ##It returns matched pattern val
```

```python
## match(pat, string)
string= "The sky is blue and Ocean is blue"
res=re.match("[A-Z]", string)
print(res)
print(res.span()) ##it returns the match positions
print(res.group()) ##It returns matched pattern val
```

```python
## matc(pat, string)
try:
    string= "The sky is blue and Ocean is blue"
    res=re.match("z", string)
    print(res)
    print(res.span()) ##it returns the match positions
    print(res.group()) ##It returns matched pattern val
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
# match(pat, string)
try:
    string = "The sky is blue and Ocean is blue"
    res = re.match("[z]+", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
## search(pat, string)
## search 5 length word
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.search("[a-zA-Z]{5}", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
###Get the words from string which length is morethan 3
##findall(pat, string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.findall("[a-zA-Z]{3,}", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
###Get the words from string which length is morethan 3
##findall(pat, string)
import re
try:
    string = "The sky is blue and Oce457an is b123lue"
    res = re.findall("[a-zA-Z]+", string)
    print(res)
   
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
##Validate mobile number
## All are digits, length of number is 10, Starts with 6,7,8,9, Remaining are 0-9
import re
try:
    mobile=input()
    res = re.findall("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is Invalid")
   
except Exception as e:
    
    print("There is no pattern matched:", e)
```

```python
##PAN Number Validation
##it allow upper and digits, length is 10, first 5 are upper, next 4 are digits,a dn last ends with upper
import re
try:
    pan=input()
    res = re.findall("^[A-Z]{5}[0-9]{4}[A-Z]$", pan)
    print(res)
    if res:
        print(f"{pan} is valid")
    else:
        print(f"{pan} is Invalid")
   
except Exception as e:
    
    print("There is no pattern matched:", e)
```
