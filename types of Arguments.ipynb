```python
### Types of Arguments
# 1.Positional Arguments: The arguments are assign to parameters as per the positional order.

def sub(x:int, y:int)->int:
    c=x-y
    return c
##Function calling Block 
a,b=map(int, input().split())
res=sub(a,b)
print(res)
print(sub(b,a))
```

```python
## 2. Keyword Arguments: Here arguments passed along with the parameters

def sub(x:int, y:int)->int:
    c=x-y
    return c
##Function calling Block 
a,b=10, 20
print(sub(x=a,y=b))
print(sub(y=b,x=a))
```

```python

def add(x:int, y:int)->int:
    c=x+y
    return c
##Function calling Block 
a,b=10, 20
print(add(x=a,y=b))
print(add(y=b,x=a))
```

```python
def add(x:int, y:int,z:int)->int:
    d=x+y+z
    return d
##Function calling Block 
a,b,c=10,20,5
res=add(a,b,c)
print(res)
print(add(b,a,c))
```

```python
### 3.Default Arguments:The parameters defined with some default values, and it assing the argument
### value if you pass the vlues for default parameters otherwise,it will assign default values.

def add(a,b,c=0,d=0):
    res=a+b+c+d
    return res
print(add(10,20))
print(add(10,20,5))
print(add(10,20,2,1))
```

```python
def mul(a,b,c=1,d=1):
    res=a*b*c*d
    return res
print(mul(1,2))
print(mul(1,2,3))
print(mul(1,2,3,4))
```

```python
### $. Variable Length Of Positional Arguments
def add(a,b=0,*args):
    print(type(args))
    print(args)
    s=0
    for num in args:
        s+=num
    return s+a+b
print(add(1))
print(add(1,2))
## print(add(1,2,b=3))
print(add(1,2,3,5,5))
print(add(1,2,3,4,5,6,7,8,9))
##print(add(a=1,2,3,4,5,6,7,8,9))
                                 
###SyntaxError: positional argument follows keyword argument
```
