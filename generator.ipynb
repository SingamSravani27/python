```python
##generator defination
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
###print(next(res))StopIteration:
```

```python
##To overcome the stop itertion error by using loops
##generator defination
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
for i in res:
    print(i)
```

```python
##return 1 to n numbers using functions and generetors
def print_nums(n):
    for i in range(1,n+1):
        print(i)
print_nums(10)
```

```python
def fun(n):
    res=[]
    for i in range(1,n+1):
        res.append(i)
        return res
print(fun(10))
```
