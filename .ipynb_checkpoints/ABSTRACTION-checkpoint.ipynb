## Abstraction


```python
from abc import ABC, abstractmethod

class Mobile(ABC):
    @abstractmethod
    def rom(self):
        pass
    @abstractmethod
    def ram(self):
        pass
    @abstractmethod
    def battery(self):
        pass

# concrete method
    def about_device(self):
        return "This is mobile"
class Oppo(Mobile):
    def __init__(self, model, oppo_ram, oppo_rom, oppo_battery):
        self.model = model
        self.oppo_ram = oppo_ram
        self.oppo_rom = oppo_rom
        self.oppo_battery = oppo_battery

    def ram(self):
        return f"The Oppo {self.model} have {self.oppo_ram} GB RAM"
        
    def rom(self):
        return f"The Oppo {self.model} have {self.oppo_rom} GB ROM"

    def battery(self):
        return f"The Oppo {self.model} have {self.oppo_battery} MAH Battery"

    def camera(self, front, back):
        return f"The Oppo {sel.model} have {front} MP front camera & {back} MP back camera"

#m = Mobile() ###Can't instantiate abstract class
reno16=Oppo("reon16",16,256,5000)
reno15=Oppo("reno15", 8, 256, 5500)
print(reno16.about_device())
print(reno16.ram())
print(reno16.rom())
```
