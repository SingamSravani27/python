## Polymorphism


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"Name: {self.name}, Age: {self.age}"
class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no
    # Overriding the details() method
    def details(self):
        return f"Name: {self.name}, Age: {self.age}, Roll No: {self.roll_no}"
# Objects
p = Person("Sravani", 21)
s = Student("Akki", 20, 101)
print(p.details())
print(s.details())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f" My Name is {self.name} and i am {self.age} years old"
class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no
    # Overriding the details() method
    def details(self):
        return f"I am student"
# Objects
p = Person("Sravani", 21)
s = Student("Akki", 20, 101)
print(p.details())
print(s.details())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def details(self):
        return f"My Name is {self.name} and I am {self.age} years old."

class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no

    # Overriding the details() method
    def details(self):
        return super().details() + f" I am a student"

# Objects
p = Person("Sravani", 21)
s = Student("Akki", 20, 101)

print(p.details())
print(s.details())
```

```python
class Math:
    def add(self,x,y):
        return a+b
    def add(self,x,y,z):
        return x+y+z
        
    def add(self,a,b,c,d):
       return a+b+c+d
m=Math()
print(m.add(1,2,3,4))
#print(m.add(1,2,3))
print(m.add(1,2))
        
```

```python
###Python directly doesn't supports method overloading,but we can achieve through the default arguments or Arbitary arguments.
##Method Overloading
class Math:
    def add(self,a,b, c=0,d=0):
     return a+b+c+d
m=Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

## Operator Overloading

```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
p1=Point(5,3)
print(p1)

        
```

```python
class Point:
    def __init__(self,x,y): ###initalize magic method
        self.x=x
        self.y=y
    def __str__(self): ###string magic method
        return f"This point class and value is:({self.x}, {self.y})"
p1=Point(5,3)
print(p1)

        
```

```python
class Point:
    def __init__(self,x,): ###initalize magic method
        self.x=x
    def __str__(self): ###string magic method
        return f"This point class and value is:({self.x})"
p1=Point(5)
p2=Point(3)
print(p1.x+p2.x)
print(p1+p2)
        
```

```python
class Point:
    def __init__(self,x): ###initalize magic method
        self.x=x
    def __str__(self): ###string magic method
        return f"This point class and value is:({self.x})"
    def __add__(self,other):
        return self.x + other.x
p1=Point(5)
p2=Point(3)
print(p1.x + p2.x)
print(p1 + p2)
###print(p1 - p2)
        
```

```python
class Point:
    def __init__(self,x): ###initalize magic method
        self.x=x
    def __str__(self): ###string magic method
        return f"This point class and value is:({self.x})"
    def __add__(self,other):
        return self.x + other.x
    def __sub__(self,other):
        return self.x - other.x
p1=Point(5)
p2=Point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1 - p2)
        
```

```python
class Point:
    def __init__(self,x): ###initalize magic method
        self.x=x
    def __str__(self): ###string magic method
        return f"This point class and value is:({self.x})"
    def __add__(self,other):
        return str(self.x) + str(other.x)
    def __sub__(self,other):
        return self.x - other.x
p1=Point(5)
p2=Point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1 - p2)
        
```
