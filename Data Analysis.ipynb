```python
## data  transformaton
categorical_columns=['mainroad','guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']
def transfer(x):
    if x=='yes':
        return 1
    else:
        return 0
        
df['mainroad']=df['mainroad'].apply(transfer)
print(df['mainroad'])
```

```python
def transfer(x):
    if x=='yes':
        return 1
    else:
        return 0
df['basement']=df['basement'].apply(transfer)
df['guestroom']=df['guestroom'].apply(transfer)
df['hotwaterheating']=df['hotwaterheating'].apply(transfer)
df['airconditioning']=df['airconditioning'].apply(transfer)
df['prefarea']=df['prefarea'].apply(transfer)

print(df['mainroad'])
print(df['guestroom'])
print(df['hotwaterheating'])
print(df['airconditioning'])
```

```python
def transfer(x):
    if x=='yes':
        return 1
    else:
        return 0
df['furnishingstatus']=df['furnishingstatus'].apply(transfer)
print(df['furnishingstatus'])
```

```python
###find Uniques values
df.furnishingstatus.value_counts()
```

```python
def transfer(x):
    if x=='unfurnished':
        return 0
    elif x=='semidurnished':
        return 0.5
    else:
        return  1
df['furnishingstatus']=df['furnishingstatus'].apply(transfer)
df
```

```python
##single 
plt.figure(figsize=(8,8))   ### pass width and height as tuple
sns.countplot(data=df, x="parking")
plt.show()
```

```python
plt.figure(figsize=(8,8))   ### pass width and height as tuple
sns.countplot(data=df, x="mainroad",hue="parking")
plt.show()
```
