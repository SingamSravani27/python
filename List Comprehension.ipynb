```python
## List Comprehension
### Syntax: [exp for iterator in iterable if condition]
## return 1-n numbers as list Using LIST COMPREHENSION
n=int(input())
res=[x for x in range(1,n+1)]
print(res)
     
```

```python
## Find the First N even numbers
n=int(input())
res=[2*x for x in range(n)]
print(res)
```

```python
n=int(input())
res=[x for x in range(0,2*n,1) ]
print(res)
```

```python
n=int(input())
res=[x for x in range(0,2*n,1) if x%2==0]
print(res)
```

```python
### Return list  even or odd, for list of intergers
arr=[20,34,33,55,23,54]
res=["Odd" if x&1 else "Even" for x in arr]
print(res)
```

```python
arr=[20,34,33,55,23,54]
res=[x+0.5 if x&1 else x+1 for x in arr]
print(res)
```

```python
###Find on dice combinations
print([x for x in range(1,7)])
```

```python
print([(i,j) for i in range(1,7) for j in range(1,7)])
```

```python
##Find two dices combinations if ny dice face is 4
print([(i,j) for i in range(1,7) for j in range(1,7) if i==4 or j==4])
```
