```python
## 1.Single Inheritance
# with Implementation in child class
##when child don't have constructor, In this case parent class constructor behaves like child class constructor
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def details(self):
        return f"My name is {self.name} and I am {self.age} years old and {self.gender}"


# Child class
class Student(Person):
    def __init__(self, stdid, name, age, gender):
        super().__init__(name, age, gender)   # Call parent constructor
        self.stdid = stdid


p = Person("babu", 21, "Male")
s = Student(101, "sravani", 20, "Female")

print(s.name)
print(p.details())

s.gender = "Male" ### adding new instance attribute
print(s.details())
```

```python
## Inherite the parent class constructor in chld class by using super()
##Super(): This function is user to get the prent clss implememtation
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old and {self.gender}"
# Child class
class Student(Person):
    def __init__(self, stdid, name, age, gender):
        super().__init__(name, age, gender)   # Call parent constructor
        self.stdid = stdid
        super().__init__(name, age, gender)
        
    def percentage(self,marks:list):
        return sum(marks)/len(marks)
        
s = Student(101, "sravani", 20, "Female")

print(s.name)
print(s.details())
print(s.percentage([90,89,88]))
```

```python
# Parent Class
class Employee:
    company = "SINGAM SRAVANI"   # Class Attribute
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary
        
# Child Class - Team Lead
class TeamLead(Employee):
    def __init__(self, emp_id, name, salary, project):
        super().__init__(emp_id, name, salary)
        self.project = project
    def calculate_salary(self,bonus=0):
        return self.salary+bonus
    def project_status(self,status):
        return f"{self.project} project status:{status}% completed"
        
class Manager(TeamLead):
    def __init__(self, emp_id, name, salary, project, no_projects):
        super().__init__(emp_id, name, salary, project)
        self.no_projects = no_projects
   
# Creating Objects
e = Employee(101, "Radha", 45000)
t = TeamLead(102, "Akki", 46000, "Banking App")
m = Manager(103, "Sravani", 76600, "AI Project", 5)
print(e.name, t.name, m.name)
print(t.calculate_salary())
print(t.calculate_salary(5000))
print(t.project_status(90))
print(m.calculate_salary(10000))
print(m.project_status(80))
```

```python
# MULTIPLE INHERITANCE

class Father:
    def __init__(self, father_name):
        self.father_name = father_name

    def working(self):
        return "Father working as Engineer"


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name

    def working(self):
        return "Mother working as Teacher"


class Child(Father, Mother):   # Mother (capital M)
    def __init__(self, name, father_name, mother_name):
        self.name = name
        Mother.__init__(self, mother_name)
        Father.__init__(self, father_name)

    def working(self):
        f = Father(self.father_name)
        m = Mother(self.mother_name)
        print(f.working())      # No ()
        print(m.working())      # No ()


c = Child("Sravani", "Radha", "Manyam")

print(c.father_name)
print(c.mother_name)

print(Child.__mro__)      # Method Resolution Order

c.working()               # Don't use print(c.working())
```

```python
class person:
    def __init__(self, name):
        self.name = name

    def details(self):
        return f"Name: {self.name}"


class student(person):
    def __init__(self, name, stdid):
        person.__init__(self, name)
        self.stdid = stdid

    def study(self):
        return "Studying"


class employee(person):
    def __init__(self, name, empid):
        person.__init__(self, name)
        self.empid = empid

    def work(self):
        return "Working"


class intern(student, employee):
    def __init__(self, name, stdid, empid):
        student.__init__(self, name, stdid)
        employee.__init__(self, name, empid)

    def info(self):
        print(self.details())
        print(self.study())
        print(self.work())


i = intern("Sravani", 10, 21)

print(i.stdid)
print(i.empid)
i.info()
```

```python
class person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"Name: {self.name}, Age: {self.age}"
class student(person):
    def __init__(self, name, age, stdid):
        self.stdid = stdid
        super().__init__(name, age)
    def study(self):
        return "Studying Python"
class employee(person):
    def __init__(self, name, age, empid):
        self.empid = empid
        super().__init__(name, age)

    def work(self):
        return "Working in Google"
s = student("Radha", 21, 101)
e = employee("Venky", 25, 201)
print(s.details())
print(s.study())
print(e.details())
print(e.work())
```
