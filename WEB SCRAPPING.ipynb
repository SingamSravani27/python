```python
import requests
from bs4 import BeautifulSoup as bs
search_ele = input().replace(" ", " ")
url = "https://www.flipkart.com/search?q=" + search_ele
try:

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/151.0.0.0 Safari/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9"
    }
    content = requests.get(url, headers=headers)
    print(content)
    web_page = bs(content.text, "html.parser")
    products = web_page.find_all(
        "div",
        {"class": "lvJbLV col-12-12"}
    )
    for product in products:
        # Product link
        try:
            product_link = "https://www.flipkart.com" + product.div.div.div.a["href"]
        except:
            product_link = "No product link"
        # Product name
        try:
            product_name = product.find(
                "div",
                {"class": "RG5Slk"}
            ).text
        except:
            product_name = "No product name"
        # Rating
        try:
            product_rating = product.find(
                "div",
                {"class": "MKiFS6"}
            ).text
        except:
            product_rating = "No product rating"
        # Discount price
        try:
            price = product.find(
                "div",
                {"class": "hZ3P6w DeU9vF"}
            ).text
        except:
            price = "No discount price"
        # Actual price
        try:
            actual_price = product.find(
                "div",
                {"class": "QiMO5r"}
            ).text
        except:
            actual_price = "No actual price"
        # Product description
        try:
            product_desc = []

            description = product.find_all(
                "li",
                {"class": "DTBslk"}
            )
            for desc in description:
                product_desc.append(desc.text)
        except:
            product_desc = "No product description"
        # Product information
        product_info = {
            "Product Name": product_name,
            "Actual Price": actual_price,
            "Discount Price": price,
            "Rating": product_rating,
            "Product Description": product_desc,
            "Product link": product_link
        }
        print(product_info)
        print("*" * 60)
except:
    print("Not able to scrap website")
```
