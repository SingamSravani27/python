```python
## Numpy array operations
arr1 =np.array([1,2,3,4])
print(arr1)
print(np.ndim(arr1)) ##It returns dimensional of an array
print(np.size(arr1)) ##It returns size of an array
```

```python
## Numpy array operations
arr1 =np.array([[1,2,3,4],[5,6,7,8]])
print(arr1)
print(np.ndim(arr1)) ## It returns dimensional of an array
print(np.size(arr1)) ## It returns total elements in side of an array
print(np.shape(arr1)) ## it returns array shape 
```

```python
arr1=np.array([1,2,3,4])
arr2=np.array([4,3,2,1])
print(arr1+arr2)
```

```python
arr1=np.array([1,2,3,4])
arr2=np.array([4,3,2])
print(arr1+arr2)
```

```python
arr1=np.array([1,2,3,4])
arr2=np.array([4,3,2,1])
print(arr1+arr2)
print(arr1-arr2)
print(np.shape(arr1),np.shape(arr2))
print(arr1+5)
print(arr1*2)
```
