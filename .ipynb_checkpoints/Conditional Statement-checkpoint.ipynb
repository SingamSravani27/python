```python
### check whether the number is divisible by 3 and 5 or not
num=int(input("enter a number"))
if num %3==0 and num %5==0:
    print(f"{num} divisible with both 3 and 5")
else:
    print(f"{num} not divisible with both 3 and 5")
```

```python
### check whether the number is perfect square or not
num=int(input("enter a value"))
root=int(num**0.5)
if root**root==num:
    print("perfect square")
else:
    print("not perfect square")
```

```python
num=int(input())
if num>=0 and num %2==0:
    print(f"{num} is +ve Even")
elif num>=0 and num %2==1:
    print(f"{num} is +ve odd")
elif num<0 and num %2==0:
    print(f"{num} is -ve Even")
elif num<0:
    print(f"{num} is -ve odd")
else:
    print("Zero")
```
