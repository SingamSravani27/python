### SNAKE AND LADDER GAME

```python
# snake and ladder
import random
snakes_and_ladders = {
    #snakes
    99: 1,
    23: 5,
    43: 9,
    56: 25,
    68: 38,
    83: 66,
    #ladders
    5: 25,
    15: 21,
    28: 42,
    50: 89,
    67: 72,
}
position = 0
rolls = 0

while True:
    input("To roll dice click enter button")
    dice_val = random.randint(1,6)
    rolls += 1
    print(f"You rolled a {dice_val}")

    if position + dice_val <= 100:
        position += dice_val

        if position in snakes_and_ladders:
            new_position = snakes_and_ladders[position]
            if new_position > position:
                print(f" Ladder! You climbed from {position} to {new_position}")
            else:
                print(f" Snake! You slid from {position} to {new_position}")
                
        if position == 100:
            print(f" You reached 100 in {rolls} rolls!")
            break

        print(f" For {rolls} rolls, you are at {position} position")
```

```python
import random

steps = 0
dice_rolls = 0

while steps < 100:
    roll = random.randint(1, 6)  
    steps += roll                
    dice_rolls += 1              # Count the roll

    print(f"Roll: {roll}, Steps: {steps}")

print(f"\nReached {steps} steps!")
print(f"Total dice rolls: {dice_rolls}")
```

```python
# Who will reach the top in fewer steps?
import random
steps = 0
rolls = 0

while True:
    input("To roll dice for that click enter button")
    dice_val = random.randint(1, 6)
    rolls += 1

    print("Current dice value:", dice_val)
    if steps + dice_val <= 100:
        steps += dice_val
        
    if steps == 100:
        print(" You reached the TOP!")
        print(f"For {rolls} rolls, you reached position {steps}")
        break
    print(f"For {rolls} rolls, you reached position {steps}")
    
```
