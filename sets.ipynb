```python
## Set Methods
-Adding Element To Set
 -add(ele): It adds element to set if element not exists in set, otherwise nothing.
 -update(iterable): It adds all iterable elements to set.
-Remove elements from set
 -remove(ele): It removes the element from set if element exists in set otherwise `Key Error`.
 -discard(ele):  It removes the element from set if element exists in set otherwise nothing.
 -pop(): It removes and returns the elements from set.

 - SET OPERATIONS
     -union(antor_sets): It returns the all common elements from all sets.
     -intersection(anthor_sets): It returns the common elements from the all sets.
     -difference(anthor_sets): It returns the set elements those are not lies in anthor sets.
     -Symmetric_difference(anthor_sets): It returns the Non-Common elements from all sets.
```

```python
s1={1,23,3,4,5,6}
s2={7,8,9,2,23,4}
s3={5,6,4}
s4=(3,5,1,34)
print(s1.union(s2))
print(s1.union(s2,s3))
print(s1.intersection(s2))
print(s2.intersection(s3))
print(s2.intersection(s1))
print(s1.difference(s2,s3))
print(s1.symmetric_difference(s2),s1.symmetric_difference(s3) ,s1.symmetric_difference(s4))
```

```python
s={1,2,3,-5}
s.add(4)
print(s)
s.add(1)   ###1 doesnot add to set because set contain 1
print(s)
s.update([10,20,30])
print(s)
s.remove(1)
print(s)
## s.remove(1)  ### Key Error
s.discard(1)
print(s)
```

```python
s1={1,23,3,4,5,6}
s2={7,8,9,2,23,4}
s3={5,6,4}
s4=(3,5,1,34)
print(s1 | s2) ###  '|'- Union 
print(s1 | s2 | s3)
print(s1 & s2)### '&'- intersection
print(s2 & s3 )
print(s2 - s1)    ### '-' for difference
print(s1 - s2 - s3)
print(s1 ^ s2)
```
