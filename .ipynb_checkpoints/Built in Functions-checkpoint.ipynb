# Built in Functions:The function defination id provided by python ,those functions called ad Built in Functions

```python
##abs():It returns the absalute value
print(abs(20-10))
print(abs(10-20))
```

```python
### round(val,pos):This function rounds the float value to particular positions 
print(round(2.2589,2))
print(round(2.2539,2))
print(round(2.2539))
print(int(2.2539))
          
```

```python
###ord(char): It returns ASCII value of character
##chr(ascii_val):It returns the charater for specified ASCII value
print(ord("A"))
print(chr(65))
```

```python
###sorted(iterable, reverse=False, key=function):It returns new sorted iterable
##sort the fruits based on length of fruit name
fruits=['Apple','banana', 'kiwi','Mango','Water Melon']
res=sorted(fruits)
print(res)
print(sorted(fruits,reverse=True))
print(sorted(fruits,reverse=True,key=len))###it sorts the list based on the length of each element or item
```

```python
##map(function, iterable)
##find the square of all list of intergers using map function
def square(x):
    return x**2
arr=[1,2,3,4,5]
res=list(map(square,arr))
print(res)
#filter(function,iterable)
##Find even numbers from list of numbers
```

```python
#filter(function,iterable): It filters the iterable, if the function returns non-none value, those values will aded to result
##Find even numbers from list of numbers
def check_even(x):
    if x%2==0:
        return x
lst=[1,2,3,4,5,6]
res=list(filter(check_even,lst))
print(res)
```

```python
def check_even(x):
    if x%2==0:
        return x
lst=[1,2,3,4,5,6]
res=list(map(check_even,lst))
print(res)
```

```python
###find the three's multipliers from list of integers
def threes_multiplers(x):
    if x%3==0:
        return x
arr=[1,24,54,22,44,51,33,15]
res=list(filter(threes_multiplers,arr))
print(res)
```

```python
###enumerate(iterable):This function returns the list of tuple of index and item for iterable elements
fruits=['Apple','banana', 'kiwi','Mango','Water Melon']
print(enumerate(fruits))
print(list(enumerate(fruits)))
for ind,val in enumerate(fruits):
    print(f"{ind}-{val} and length id{len(val)}")
    
```

```python
###Tuple unpacking
t=(1,2)
a,b=t
print(a,b)
```

```python
###Zip(iterable1,iterable2,....):It returns all iterable as singlr iterable of tuples
rool_no=[101,102,103,104,105,106]
names=['Ravi','Raju','Siva','Rani','Gopi']
percentage=[90,67,96.8,67.65,99.99999]
print(list(zip(rool_no,names,percentage)))
```

```python
### Number Conversion Function
num=int(input())
res=""
while num>0:
    bit=num%2
    num=num//2
    res=str(bit)+res
print(res)
```

```python
print(bin(10))###string type of binary value
print(bin(10)[2:])
print(oct(10))
print(hex(11).upper())
print(hex(11).upper()[2:])
```

```python

###decimal to hexadecimal
num=int(input())
res=""
while num>0:
    bit=num%16
    num=num//16
    res=str(bit)+res
print(res)
```

```python
##any(): This function returns True when any on eiterable items return True
##all():Returns True if bool(x) is True for all values x in the iterable
arr=[1,2,3,4,0]
print(all(arr))
print(any(arr))
```

```python
percentage=[34,34,54,65,67]
for num in percentage:
    if num<35:
        print("All students not passed")
        break
    else:
        print("All students pass")
```

```python
percentage=[34,34,54,65,67]
lst=[]
for num in percentage:
    if num<35:
        lst.append(False)
    else:
        lst.append(True)
print(lst)
if all(lst):
    print("All students passed")
else:
    print("All students not pass")
```
