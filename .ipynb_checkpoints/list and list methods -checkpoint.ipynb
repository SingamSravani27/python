```python
##List ordered collection
lst=[1,2,4,5.65,'sravani','codegnan']
print(lst)
```

```python
#List is mutable
lst=[1,2,4,5.65,'sravani','codegnan']
lst[2]=100
print(lst)
```

```python
#Indexing and slicing
lst=[1,2,4,5.65,'srinu','codegnan']
print(lst[-2])
print(lst[-1])
print(lst[-3:])
print(lst[:100])
print(lst[:100:10])
print(lst[-10:-1:-1])
print(lst[-10:-1])
print(lst[::-1])
print(lst[::-2])
print(lst[:-1:-3])
```

```python
## list operators


# list concatination (using '+' operator)
l1=[1,2,3,4,]
l2=[4,5,6]
res=l1+l2
print(res)
print(l1+l2)

# List repetation (using '*' operator)
print(l1*3)
print([0]*10)
print([]*10)
```

```python
## add single element 10 to list
l1=[1,2,3,4]
res=l1+[10]
print(res)
```

```python
## List Methods
- Element Adding Methods
 -'append(item)': It adds items at end of the list
 -'extend(itarable)':It is used to add all iterable elements at the end of the list
 -'insert(pos,item)': This method adds the item t specified position
```

```python
lst=[1,2,3,4]
print(lst.append(10))
print(lst)

lst.append('Code')
print(lst)

lst.append([10,20,30])
print(lst)
print(lst[6])
print(lst[6][2])

##extend
lst.extend([100,200,300])
print(lst)

##insert
lst.insert(1,2000)
print(lst)
```

```python
-Element Removing Methods
 - remove(val): It removes first occurance of the vlue if value exits in list, otherwise it raises 'value error'
 -pop(): It removes and returns the last element of list.
 -pop(pos): It removes and returns specified position element from list
 -clear(): It removes all element from list
 -del  : It is python keyword, used to delete values.
```

```python
lst=[10,20,30,40]
lst.remove(20)
print(lst)
#lst.remove(100)   #ValueError: list.remove(x): x not in list

val=lst.pop()
print(val,lst)
val2=lst.pop(1)   # it removes the lst postiton element
print(val2, lst)
#print(lst.pop(10))  #IndexError: pop index out of range

del lst[0]
print(lst)

lst.clear()
print(lst)
```

```python
# Index(val): It returns the index of first occurance of value if value exists in list otherwise ValueError.

## count(val): Itreturns the total frequence of value

lst=[11,2,12,4,14,2,2,2]
ind=lst.index(2)
print(ind)
print(lst.count(2))
print(lst.count(100))
##print(lst.index(100))  ###ValueError: list.index(x): x not in list
```

```python
## sort(reverse= false): It sorts he all list elements the all list elements in Ascending order, 
    ##                if reverse value is true it sorts in Descending order.

lst=[5,43,63,6,73,-2,4,-6]
lst.sort()
print(lst)

### Reversing
lst.sort(reverse=True)
print(lst)
```

```python
# Built in functions
#len(), min(), max(),sum()
lst=[1,2,3,4]
print(len(lst),min(lst),max(lst),sum(lst))

lst=['Hello','hi','where','are','you']
print(len(lst),min(lst),max(lst))

lst.sort()
print(lst)
```
