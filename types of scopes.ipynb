```python
##Enclosing Scope
def outer(x):
    y=10 ###Enclosing scope
    def inner():
        a=10#local scope
        z=y+a
        print(f"This is local scope:{z}")
    inner()
    y=y+x
    print(f"this is enclosing scope:{y}")
n=5
outer(n)
```

```python
##Enclosing Scope
def outer(x):
    x=10 ###Enclosing scope
    def inner():
        x=20#local scope
        z=x+x
        print(f"This is local scope:{z}")
    inner()
    y=x+x
    print(f"this is enclosing scope:{y}")
n=5
outer(n)
    
```

```python
###Global scope
x=10  #Global scope
def outer():
    y=20 #enclosing scope
    def inner():
        z=30 #local scope
        res=z+y+x
        print(f"this is local scope:{res}")
    res=y+x
    inner()
    print(f"this is enclosing scope:{res}")
outer()
        
```

```python
## trying to upadte the global scope variable inside the local scope
x=10  #global scope
def outer():
    x=20
    x=x+1
    print("this is local val:",x)
outer()
print("this is global scope val:",x)
```

```python
##update the global scope variable inside the local scope by using 'global' keyword
x=10  #global scope
def outer():
    global x
    x=x+1
    print("this is local val:",x)
outer()
print("this is global scope val:",x)
```

```python
x=10  #Global scope
def outer():
    y=20 #enclosing scope
    def inner():
        nonlocal y
        z=30 #local scope
        y=z+1
        print(f"this is local scope:{y}")
    y=y+x
    inner()
    y=y+x
    print(f"this is enclosing scope:{y}")
outer()
```
