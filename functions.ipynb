```python
### find Factors of a number
num=int(input())
factors=[]
for i in range(1,num+1):
    if num%i==0:
        factors.append(i)
print(factors)
if len(factors) ==2:
    print(f"{num} is prime")
else:
    print(f"{num} is not prime")
```

```python
string=input()
rev=""
for char in string:
    rev=char+rev
    print(rev)
if rev==string:
        print(f"{string} is palindrome")
else:
        print(f"{string} is not palindrome")
    
```
