```python
##define empty tuple
t1=()
t2=tuple()
print(type(t1),type(t2))
```

```python
l1=[10] 
t1=(10)                                         ###() groups expressions and controls the order of operations.
t2=(10, )                                        ###() does not automatically create a tuple.
print(type(l1),type(t1),type(t2))                ### A comma , creates a tuple.
```

```python
### Dictionary 
##details={'name':'Sravani', 'age':22, 'course':'python',[1,2,3]:10}   ####   TypeError: cannot use 'list' as a dict key (unhashable type: 'list')
details={'name':'Sravani', 'age':22, 'course':'python'}
print(details)
```

```python
 ### Dictionary can't allow list as key
## If dictionary contains multiple items wiyh key, In that cse it updates first item key value with last defined item value
details={'name':'Sravani', 'age':22, 'course':'python','marks':[91,82,73],'age':50} ###update with lastest item defined value
print(details)
```

```python
### Access values by using keys
details={'name':'Sravani', 'age':22, 'course':'python','marks':[91,82,73],'age':50}
print(details['course'])
print(details['age'])


##Updating the Dictionary Value
details['age']+=1
details['course']='java'  ## It updates the value for 'course' key
print(details)

## item Assignment
details['location']='India'  ##It adds new item to dictionary
print(details)
```

```python
### Define Empty Dictionary
d1={}
d2=dict()
print(type(d1),type(d2))
```

```python
### Dictionary methods
-Get value
 -get(key,default_val):  It returns the value if key exists in diictionary otherwise `default_Val`
 -Update values and ad items to dictionary
  update(new_dictionary):Dictionary updtes with New_dictionary Items
 -Remove items from dictionary
  -pop(key): It returns value and remove item if key is exists in dictionary oyherwise `KeyError`
  -popitem(): It returns tuple of key value pair and removes the last item from dictionary
  -Get Keys,value and items from dictionary
  -Keys(): It returns list of dictionary
  -Values(): It returns list of dictionary values
  -Items(): It returns list of tuple key-value pairs
```

```python
details={'name':'Sravani', 'age':22, 'course':'python','marks':[91,82,73],'age':50}
##pritn(details['location'])
print(details.get('name'))
print(details.get('location'))
print(details.get('age',18))
print(details.get('location',"location key not exists"))
```

```python
details={'name':'Sravani', 'age':22, 'course':'python'}
d={'marks':[91,82,73],'age':50}
##update
details.update(d)
print(details)
```

```python
details={'name':'Sravani', 'age':22, 'course':'python','marks':[91,82,73],'age':50}
print(details)
print(details.pop('age'))
print(details)
## pop item
print(details.popitem())
print(details)
##print(details.pop('percentage'))  ###KeyError: 'percentage'
print(details.values())
print(details.items())
      
```

```python
details={'name':'Sravani', 'age':22, 'course':'python','marks':[91,82,73],'age':50}
keys=details.keys()
print(type(keys))
print(keys)
print(details.values())
print(details.items())
print(list(keys))
```
