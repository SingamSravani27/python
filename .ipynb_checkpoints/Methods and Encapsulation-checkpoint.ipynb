```python
#create mobile class
class Mobile:
    company="Iphone"
    def __init__(self, model, ram, rom, price):
        self.model=model
        self.ram=ram
        self.rom=rom
        self.price=price
    def about_phone(self):
        return f"The{self.model} have {self.ram} GD RAM and {self.rom}GB ROM"
    def display():
        return "This is samsung company display"
    #class Method
    @classmethod
    def update_company(cls, new_name):
        cls.company=new_name
        return "Company name updates"
     #static method
    def charger_provide(cable=True):
        return "adapter+cable" if cable else "only adapter" 
        
#object creation
iphone16=Mobile("iphone 16", 16, 256, 125000)
iphone17=Mobile("iphone 17", 16, 256, 120000)
print(iphone16.about_phone())
print(iphone16.update_company("Iphone Pro"))
print(iphone16.charger_provide())
print(iphone17.company)
```

```python
class Account:

    def __init__(self, name, balance, password):
        self.name = name             # Public
        self._balance = balance      # Protected
        self.__password = password   # Private

    # Getter method
    def get_balance(self):
        return self._balance

    # Withdraw method
    def withdraw(self, amount):
        if self._balance >= amount:
            self._balance -= amount
            return f"{amount} withdrawn successfully"
        return "Insufficient Balance!"

    # Deposit method
    def deposit(self, amount):
        self._balance += amount
        return f"{amount} deposited successfully. Current Balance: {self._balance}"

    # Private method
    def __total_users(self):
        return "Total users: 100"
# Object creation
user = Account("sravani", 5000, "sravani@123")

print(user.name)
print(user.get_balance())
print(user.deposit(5000))
```

```python
class Account:

    def __init__(self, name, balance, password):
        self.name = name             # Public
        self._balance = balance      # Protected
        self.__password = password   # Private

    # Getter method
    def get_balance(self):
        return self._balance

    # Withdraw method
    def withdraw(self, amount):
        if self._balance >= amount:
            self._balance -= amount
            return f"{amount} withdrawn successfully"
        return "Insufficient Balance!"

    # Deposit method
    def deposit(self, amount):
        self._balance += amount
        return f"{amount} deposited successfully. Current Balance: {self._balance}"

    # Private method
    def __total_users(self):
        return "Total users: 100"

    # Setter method
    def update_password(self, new_password):
        self.__password = new_password
        return "Password updated successfully"

    # Getter method
    def get_password(self):
        return self.__password


# Object creation
user = Account("sravani", 5000, "sravani@123")

print(user.name)
print(user.get_balance())
print(user.deposit(5000))

print(user.update_password("Padma123"))
print(user.get_password())

##access the protected member syntax" obj._attributename
print(user._balance)
## access private member by using name mangling
## syntax: obj._classname_attributename
print(user._Account__total_users())
```
