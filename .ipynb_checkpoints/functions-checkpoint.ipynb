```python
### find Factors of a number
num=int(input())
factors=[]
for i in range(1,num+1):
    if num%i==0:
        factors.append(i)
print(factors)
if len(factors) ==2:
    print(f"{num} is prime")
else:
    print(f"{num} is not prime")
```

```python
string=input()
rev=""
for char in string:
    rev=char+rev
    print(rev)
if rev==string:
        print(f"{string} is palindrome")
else:
        print(f"{string} is not palindrome")
    
```

```python
##FUNCTIONS
## 4. With return and with parameter
## sum of two numbers
#function defination block
def add(x:int, y:int)->int:### what data type it is executing or writtening
    c=x+y
    return c
### function calling
a, b=5, 6
res=add(a, b)
print(res)
print(add(1,2))
print(add(3,4))
```

```python
### 3.with return and without parameters
def add()->int:### what data type it is executing or writtening
    x,y=10,20
    c=x+y
    return c
### function calling
res=add()
print(res)
print(add())
print(add())
print(add())
```

```python
###find whether the number is prime number or not using function
def prime_check(n:int)->bool:
    if n<2:
        return False
    for i in range(2, n//2 + 1):
        if n % i == 0:
            return False
    return True
for j in range(1,100):
    if prime_check(j):
        print(j,end=" ")
```
