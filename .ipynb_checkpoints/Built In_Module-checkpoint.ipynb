```python
import random
print(random.random()) ##This function returns the valye in between 0 to 1
print(random.randint(1,10)) ## radiant(a,b):This function generates random number between a,b
print(random.choice(['A',1,2,3,4,5,6,7,8,9,'J','Q','K']))   ##It returns the random items from list
print(random.choices(['A',1,2,3,4,5,6,7,8,9,'J','Q','K'],k=3)) 
arr=['A',1,2,3,4,5,6,7,8,9,'J','Q','K']
random.shuffle(arr)
print(arr)
```

```python
import random

steps = 0
dice_rolls = 0

while steps < 100:
    roll = random.randint(1, 6)  
    steps += roll                
    dice_rolls += 1              # Count the roll

    print(f"Roll: {roll}, Steps: {steps}")

print(f"\nReached {steps} steps!")
print(f"Total dice rolls: {dice_rolls}")
```

```python
# Who will reach the top in fewer steps?
import random
steps = 0
rolls = 0

while True:
    input("To roll dice for that click enter button")
    dice_val = random.randint(1, 6)
    rolls += 1

    print("Current dice value:", dice_val)
    if steps + dice_val <= 100:
        steps += dice_val
        
    if steps == 100:
        print(" You reached the TOP!")
        print(f"For {rolls} rolls, you reached position {steps}")
        break
    print(f"For {rolls} rolls, you reached position {steps}")
```
