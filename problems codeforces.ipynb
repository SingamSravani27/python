```python
import math

A, B = map(int, input().split())

x = A / B

print(f"floor {A} / {B} = {math.floor(x)}")
print(f"ceil {A} / {B} = {math.ceil(x)}")
print(f"round {A} / {B} = {round(x)}")
```
